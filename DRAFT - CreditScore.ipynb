Multi-Collinearity Among Predictors (Feature Redundancy)
Multicollinearity occurs when predictor variables are highly correlated with each other. This can destabilize traditional regression models (like Logistic Regression) by inflating standard errors:

ROE and 5Yr Avg ROE (+0.88): Extremely high correlation. Including both introduces severe redundancy; typically, keeping the 5-year average is preferred to smooth out cyclical noise.

Curr Ratio and Cash Ratio (+0.66): Strong liquidity overlap. Both measure short-term cushion, though they capture slightly different aspects (inventory/receivables vs. pure cash).

Debt/Assets and Net Dbt / EBITDA Adj (+0.52): Reflects overlapping dimensions of structural and cash-flow leverage.

Implications for Your Machine Learning Models
For Ordered Logistic Regression: High multicollinearity between features like ROE and 5Yr Avg ROE explains why you encountered Hessian matrix inversion warnings (LinAlgError). Dropping redundant features or utilizing regularization ensures a stable mathematical solution.

For CART Decision Trees: Tree models are entirely immune to multicollinearity because they choose only the single best feature split at each node. However, cleaning up redundant features often makes the resulting decision tree easier to interpret.

Feature Selection & Train/Test Split

```python
#Feature Selection & Train/Test Split

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

# 1. Select predictor columns (exclude metadata and target scales)
drop_cols = [
    'Ticker',
    'Tkr & Exch',
    'Short Name',
    'S&P LT LC',
    'Rating Scale',
    'ABS Rating Scale',
    'Tier Label',
]
X = df.drop(columns=[c for c in drop_cols if c in df.columns]).copy()

# Apply log1p transform to skewed monetary metrics
for col in ['Mkt Cap', 'Tot Assets']:
    if col in X.columns:
        X[col] = np.log1p(X[col])

# One-hot encode categorical features (Sector, Country, etc.)
categorical_cols = X.select_dtypes(include=['object']).columns.tolist()
X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)

# 2. Define targets
y_granular = df['Rating Scale']  # Exp 1: Granular (Continuous / Notches)
y_broad = df['ABS Rating Scale']  # Exp 2: Broad (Ordinal Tiers 1 to 5)

# 3. Train/Test Splits (80/20)
X_train_g, X_test_g, y_train_g, y_test_g = train_test_split(
    X, y_granular, test_size=0.20, random_state=42
)

X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
    X, y_broad, test_size=0.20, random_state=42, stratify=y_broad
)

print(
    f'Data split successfully! Training: {X_train_g.shape[0]} rows, Testing: {X_test_g.shape[0]} rows'
)
```

Experiment 1 — Granular Rating Scale (CART vs. Ridge/Linear)

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.tree import DecisionTreeRegressor

# 1. Train CART Regressor (Decision Tree)
cart_reg = DecisionTreeRegressor(max_depth=5, random_state=42)
cart_reg.fit(X_train_g, y_train_g)
y_pred_cart_g = cart_reg.predict(X_test_g)

# 2. Evaluate Performance
mae_cart = mean_absolute_error(y_test_g, y_pred_cart_g)
rmse_cart = np.sqrt(mean_squared_error(y_test_g, y_pred_cart_g))
r2_cart = r2_score(y_test_g, y_pred_cart_g)

# Notch Hit Rates
hit_1_notch = np.mean(np.abs(y_pred_cart_g - y_test_g) <= 0.33) * 100
hit_2_notch = np.mean(np.abs(y_pred_cart_g - y_test_g) <= 0.66) * 100

print('=== EXPERIMENT 1: GRANULAR RATING SCALE (CART REGRESSOR) ===')
print(f'R² Score:            {r2_cart:.4f}')
print(f'Mean Absolute Error: {mae_cart:.4f}')
print(f'RMSE:                {rmse_cart:.4f}')
print(f'Within ±1 Notch:     {hit_1_notch:.2f}%')
print(f'Within ±2 Notches:   {hit_2_notch:.2f}%')
```

```python
import numpy as np
import pandas as pd

# 1. Define dictionary mapping S&P letter ratings to ordered integer ranks (1 = Best, higher = Riskier)
rating_rank_map = {
    'AAA': 1,
    'AA+': 2,
    'AA': 3,
    'AA-': 4,
    'A+': 5,
    'A': 6,
    'A-': 7,
    'BBB+': 8,
    'BBB': 9,
    'BBB-': 10,
    'BB+': 11,
    'BB': 12,
    'BB-': 13,
    'B+': 14,
    'B': 15,
    'B-': 16,
}

# 2. Check for any ratings in the data not covered by the map (e.g. 'NR', stray whitespace, lower tiers)
unmapped = df.loc[~df['S&P LT LC'].isin(rating_rank_map.keys()), 'S&P LT LC'].value_counts()
if len(unmapped) > 0:
    print('=== WARNING: Unmapped S&P LT LC values found (will be dropped) ===')
    print(unmapped)
    df = df[df['S&P LT LC'].isin(rating_rank_map.keys())].copy()

# 3. Create the new ordered column in the dataframe
df['Rating_Rank'] = df['S&P LT LC'].map(rating_rank_map)

# 4. Show a preview table verifying the mapping
print('\n=== Sample Table: Letter Rating to Ordered Rank Mapping ===')
display(df[['Ticker', 'Short Name', 'S&P LT LC', 'Rating Scale', 'Rating_Rank']].head(10))

print(f'\nRemaining rows after dropping unmapped ratings: {len(df)}')
print(f'NaNs in Rating_Rank: {df["Rating_Rank"].isna().sum()}')
display(df[['Ticker', 'Short Name', 'S&P LT LC', 'Rating Scale', 'Rating_Rank']].head(10))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import RobustScaler

# 1. Define feature matrix X and target y (using our new Rating_Rank column)
drop_cols = [
    'Ticker',
    'Tkr & Exch',
    'Short Name',
    'S&P LT LC',
    'Rating Scale',
    'ABS Rating Scale',
    'Rating_Rank',
    'Tier Label',
]
X = df.drop(columns=[c for c in drop_cols if c in df.columns]).copy()

# Apply log1p transform to skewed monetary metrics
for col in ['Mkt Cap', 'Tot Assets']:
  if col in X.columns:
    X[col] = np.log1p(X[col])

# One-hot encode categorical variables (Sector, Country, Currency, etc.)
categorical_cols = X.select_dtypes(include=['object']).columns.tolist()
X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)

y_rank = df['Rating_Rank']



# 2. Perform 80/20 Train/Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y_rank, test_size=0.20, random_state=42
)

# 3. Scale numerical data using RobustScaler (Fit on Train ONLY)
scaler = RobustScaler()
X_train_scaled = pd.DataFrame(
    scaler.fit_transform(X_train), columns=X_train.columns, index=X_train.index
)
X_test_scaled = pd.DataFrame(
    scaler.transform(X_test), columns=X_test.columns, index=X_test.index
)

print(f'Training Set: {X_train_scaled.shape[0]} rows')
print(f'Testing Set:  {X_test_scaled.shape[0]} rows')
```

```python
from statsmodels.miscmodels.ordinal_model import OrderedModel
from sklearn.preprocessing import RobustScaler

# 1. Select a clean, non-collinear subset of core S&P financial ratios
#PROBLEM WITH MULTICOLLINEARITY, SO WE NEED TO DROP VARIABLES
core_features = [
    'Mkt Cap',
    'Tot Assets',
    'Curr Ratio',
    'Debt/Assets',
    'Debt/EBITDA',
    'ROA',
    'EBITDA / Iexp',
    'FCF Margin',
    'BB 5Y Dflt Prob',
]

X_tr_clean = X_train[core_features]
X_te_clean = X_test[core_features]

# 2. Rescale clean subset
scaler = RobustScaler()
X_tr_clean_scaled = pd.DataFrame(
    scaler.fit_transform(X_tr_clean),
    columns=core_features,
    index=X_train.index,
)
X_te_clean_scaled = pd.DataFrame(
    scaler.transform(X_te_clean), columns=core_features, index=X_test.index
)

# 3. Fit Model cleanly using BFGS or Newton
ordered_model = OrderedModel(y_train, X_tr_clean_scaled, distr='logit')
ordered_results = ordered_model.fit(method='bfgs', maxiter=1000, disp=True)

# 4. Generate Predictions
prob_preds = ordered_results.predict(X_te_clean_scaled)
y_pred_ordered_rank = np.argmax(prob_preds.values, axis=1) + 1
```

Experiment 2 — Broad ABS Rating Scale (Ordered Logit vs. CART)
For the broad tiers (1 to 5), run both Ordered Logistic Regression (proportional odds model) and CART Classifier (DecisionTreeClassifier):

```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
)
from sklearn.preprocessing import RobustScaler
from sklearn.tree import DecisionTreeClassifier
from statsmodels.miscmodels.ordinal_model import OrderedModel

# Scale numeric predictors for Ordered Logistic Regression convergence
scaler = RobustScaler()
X_train_b_scaled = pd.DataFrame(
    scaler.fit_transform(X_train_b),
    columns=X_train_b.columns,
    index=X_train_b.index,
)
X_test_b_scaled = pd.DataFrame(
    scaler.transform(X_test_b), columns=X_test_b.columns, index=X_test_b.index
)

# -------------------------------------------------------------
# A. Ordered Logistic Regression (statsmodels)
# -------------------------------------------------------------
# Fit cumulative logit model
ord_logit = OrderedModel(
    y_train_b, X_train_b_scaled, distr='logit'
)  # or 'probit'
ord_res = ord_logit.fit(method='bfgs', maxiter=500, disp=False)

# Predict probabilities and assign the class with highest probability
prob_preds = ord_res.predict(X_test_b_scaled)
y_pred_ord = np.argmax(prob_preds.values, axis=1) + 1

# -------------------------------------------------------------
# B. CART Classifier (Decision Tree)
# -------------------------------------------------------------
cart_clf = DecisionTreeClassifier(
    max_depth=4, class_weight='balanced', random_state=42
)
cart_clf.fit(X_train_b, y_train_b)
y_pred_cart_b = cart_clf.predict(X_test_b)

print('=== EXPERIMENT 2: BROAD ABS RATING SCALE (TIERS 1 to 5) ===')
print(f'Ordered Logit Accuracy:    {accuracy_score(y_test_b, y_pred_ord):.4f}')
print(
    f'Ordered Logit Balanced Acc: {balanced_accuracy_score(y_test_b, y_pred_ord):.4f}'
)
print('-' * 55)
print(f'CART Classifier Accuracy:  {accuracy_score(y_test_b, y_pred_cart_b):.4f}')
print(
    f'CART Balanced Accuracy:    {balanced_accuracy_score(y_test_b, y_pred_cart_b):.4f}'
)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
)

# -------------------------------------------------------------
# 1. Calculate Confusion Matrices
# -------------------------------------------------------------
tier_names = ['AAA/AA (1)', 'A (2)', 'BBB (3)', 'BB (4)', 'B (5)']

# Ordered Logistic Regression Matrix
cm_ord = confusion_matrix(y_test_b, y_pred_ord)

# CART Classifier Matrix
cm_cart = confusion_matrix(y_test_b, y_pred_cart_b)

# -------------------------------------------------------------
# 2. Plot Confusion Matrix Heatmaps
# -------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# A. Ordered Logistic Regression Heatmap
sns.heatmap(
    cm_ord,
    annot=True,
    fmt='d',
    cmap='Blues',
    cbar=False,
    ax=axes[0],
    xticklabels=['1', '2', '3', '4', '5'],
    yticklabels=['1', '2', '3', '4', '5'],
    annot_kws={'size': 13, 'weight': 'bold'},
)
axes[0].set_title(
    'Ordered Logistic Regression\n(Confusion Matrix)',
    fontsize=13,
    fontweight='bold',
)
axes[0].set_xlabel('Predicted Rating Tier', fontsize=11)
axes[0].set_ylabel('Actual Rating Tier', fontsize=11)

# B. CART Classifier Heatmap
sns.heatmap(
    cm_cart,
    annot=True,
    fmt='d',
    cmap='Oranges',
    cbar=False,
    ax=axes[1],
    xticklabels=['1', '2', '3', '4', '5'],
    yticklabels=['1', '2', '3', '4', '5'],
    annot_kws={'size': 13, 'weight': 'bold'},
)
axes[1].set_title(
    'CART Decision Tree Classifier\n(Confusion Matrix)',
    fontsize=13,
    fontweight='bold',
)
axes[1].set_xlabel('Predicted Rating Tier', fontsize=11)
axes[1].set_ylabel('Actual Rating Tier', fontsize=11)

plt.suptitle(
    'Confusion Matrices on Test Set (1 = AAA/AA, 5 = B)',
    fontsize=15,
    fontweight='bold',
    y=1.03,
)
plt.tight_layout()
plt.show()

# -------------------------------------------------------------
# 3. Print Detailed Classification Reports
# -------------------------------------------------------------
print('=' * 65)
print('ORDERED LOGISTIC REGRESSION: CLASSIFICATION REPORT')
print('=' * 65)
print(
    classification_report(
        y_test_b, y_pred_ord, target_names=tier_names, zero_division=0
    )
)

print('=' * 65)
print('CART CLASSIFIER: CLASSIFICATION REPORT')
print('=' * 65)
print(
    classification_report(
        y_test_b, y_pred_cart_b, target_names=tier_names, zero_division=0
    )
)

# -------------------------------------------------------------
# 4. Comprehensive Metrics Comparison Table
# -------------------------------------------------------------
models = {
    'Ordered Logistic Regression': y_pred_ord,
    'CART Classifier (Tree)': y_pred_cart_b,
}

metrics_list = []

for name, preds in models.items():
  metrics_list.append({
      'Model': name,
      'Accuracy': f'{accuracy_score(y_test_b, preds) * 100:.2f}%',
      'Balanced Accuracy': (
          f'{balanced_accuracy_score(y_test_b, preds) * 100:.2f}%'
      ),
      'Macro Precision': (
          f"{precision_score(y_test_b, preds, average='macro', zero_division=0) * 100:.2f}%"
      ),
      'Macro Recall': (
          f"{recall_score(y_test_b, preds, average='macro', zero_division=0) * 100:.2f}%"
      ),
      'Macro F1-Score': (
          f"{f1_score(y_test_b, preds, average='macro', zero_division=0) * 100:.2f}%"
      ),
      'Weighted Precision': (
          f"{precision_score(y_test_b, preds, average='weighted', zero_division=0) * 100:.2f}%"
      ),
      'Weighted Recall': (
          f"{recall_score(y_test_b, preds, average='weighted', zero_division=0) * 100:.2f}%"
      ),
      'Weighted F1-Score': (
          f"{f1_score(y_test_b, preds, average='weighted', zero_division=0) * 100:.2f}%"
      ),
  })

comparison_summary = pd.DataFrame(metrics_list).set_index('Model')

print('\n=== OVERALL MODEL PERFORMANCE METRICS TABLE ===')
display(comparison_summary)
```
